```python
import urllib.request
import os
from datetime import datetime

def download_vhi_data():
    # Створюємо папку для даних, якщо її немає
    if not os.path.exists("vhi_data"):
        os.makedirs("vhi_data")

    # Список ID областей України (від 1 до 27)
    for i in range(1, 28):
        # Перевіряємо, чи вже існує файл для цієї області
        existing_files = [f for f in os.listdir("vhi_data") if f.startswith(f"vhi_id_{i}_") and f.endswith('.csv')]
        if existing_files:
            print(f"Файл для області {i} вже існує. Пропускаємо.")
            continue

        # Посилання на дані NOAA
        url = f"https://www.star.nesdis.noaa.gov/smcd/emb/vci/VH/get_provinceData.php?country=UKR&provinceID={i}&year1=1981&year2=2024&type=VHI"
        
        # Формуємо ім'я файлу
        now = datetime.now().strftime("%Y%m%d_%H%M%S")
        filename = f"vhi_id_{i}_{now}.csv"
        filepath = os.path.join("vhi_data", filename)

        try:
            # Створюємо запит з User-Agent (деякі сервери його вимагають)
            req = urllib.request.Request(
                url, 
                data=None,
                headers={
                    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
                }
            )
            
            # Завантажуємо файл
            with urllib.request.urlopen(req) as response:
                content = response.read().decode('utf-8')
                
                # Перевіряємо, чи це не HTML
                if '<html' in content.lower():
                    print(f"Помилка: Область {i} — сервер повернув HTML замість CSV")
                    # Видаляємо некоректний файл
                    if os.path.exists(filepath):
                        os.remove(filepath)
                    continue
                
                # Зберігаємо вміст у файл
                with open(filepath, 'w', encoding='utf-8') as f:
                    f.write(content)
                    
            print(f"Область {i} успішно завантажена: {filename}")
            
        except Exception as e:
            print(f"Помилка під час завантаження області {i}: {e}")
            # Видаляємо файл, якщо він створений, але порожній
            if os.path.exists(filepath):
                os.remove(filepath)

# Запускаємо завантаження
download_vhi_data()
```
